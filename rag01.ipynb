```python
import google.generativeai as genai
from langchain_google_genai import ChatGoogleGenerativeAI

import os
from IPython.display import Markdown
import textwrap
from dotenv import load_dotenv

load_dotenv()  # take environment variables
```

```python
def to_markdown(text):
  text = text.replace('•', '  *')
  return Markdown(textwrap.indent(text, '> ', predicate=lambda _: True))
```

```python
genai.configure(api_key=os.getenv('GOOGLE_API_KEY'))
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-2.0-flash",
    temperature=0,
    max_tokens=None,
    timeout=None,
    max_retries=2,
    # other params...
)
result = llm.invoke("What is the Retrieved Augmented Generation?")
```

```python
to_markdown(result.content)
```

```python
messages = [
    (
        "system",
        "Based on input taxon name, return a dictionary with the taxon name, r-strategy boolean and k-strategy boolean, such as {'k-strategy': True, 'r-strategy': False}"
        "in case the taxon is considered K- strategist. If the taxon is considered r-strategist, return {'k-strategy': False, 'r-strategy': True}"
        "If the taxon is considered both r or k strategist, return {'k-strategy': True, 'r-strategy': True}"
    ),
    ("human", "Is Candidatus_Nanopelagicaceae r or k strategist?"),
]
ai_msg = llm.invoke(messages)
```

```python
to_markdown(ai_msg.content)
```

```python
taxa = ['Pseudoalteromonadaceae',
 'Oceanospirillaceae',
 'Acartiidae',
 'Flavobacteriaceae',
 'Moraxellaceae',
 'Vibrionaceae',
 'Rhodobacteraceae',
 'Sporichthyaceae',
 'Chaetocerotaceae',
 'Gemmataceae',
 'Planococcaceae',
 'Oithonidae',
 'Oikopleuridae',
 'Serpulidae',
 'Cryomorphaceae',
 'Ilumatobacteraceae',
 'Candidatus_Actinomarinaceae',
 'Stephanodiscaceae',
]
```

```python
import time
# invoke the same mesaage for each taxon in the list
asnwer = []
for taxon in taxa:
    messages = [
        (
            "system",
            "Based on input taxon name, return a dictionary with the taxon name, r-strategy boolean and k-strategy boolean, such as {'k-strategy': True, 'r-strategy': False}"
            "in case the taxon is considered K- strategist. If the taxon is considered r-strategist, return {'k-strategy': False, 'r-strategy': True}"
            "If the taxon is considered both r or k strategist, return {'k-strategy': True, 'r-strategy': True}"
        ),
        ("human", f"Is {taxon} r or k strategist?"),
    ]
    ai_msg = llm.invoke(messages)
    print(f"{taxon}: {ai_msg.content}")

    # append the result to a list
    asnwer.append(ai_msg.content)
    time.sleep(5)  # sleep for 1 second to avoid rate limit
```

```python
# save dictionaly to a file
with open('taxa.json', 'w') as f:
    f.write(str(asnwer))
```
